```python
##If Chinese is your native language then you might want to explore this package
import jieba
text = "中文的單字邊界是否使用空格"
seg_list = jieba.cut(text, cut_all=False)
print("Default Mode: " + "/ ".join(seg_list))  # Default is precise mode
```
